### Kuder-Richardson Formula 20 (KR-20) Reliability Test on Pre-Matched Knowledge Dataset

```python
# We'll be making use of the mkb3 dataset
# 1. Number of items (k)
k_mkb = mkb3.shape[1]
print(f"Number of items (k): {k_mkb}")
```

```python
# 2. Calculate p_mkb and q_mkb for each item
# p_mkb = proportion of 'true' (1) responses for each item
# q_mkb = proportion of 'false' (0) responses for each item (1 - p_i)

p_mkb = mkb3.mean()
q_mkb = 1 - p_mkb

# Calculate sum(p_mkb * q_mkb)
sum_pq_mkb = (p_mkb * q_mkb).sum()
print(f"Sum of (p_mkb * q_mkb): {sum_pq_mkb:.1f}")
```

```python
# 3. Calculate variance of total scores (sigma_t^2)
# Calculate total score for each respondent (sum across rows)
total_scores_mkb = mkb3.sum(axis=1)

# Calculate the variance of these total scores
sigma_t_squared_mkb = np.var(total_scores_mkb)
print(f"Variance of total scores (sigma_t^2): {sigma_t_squared_mkb:.1f}")
```

```python
# 4. Apply the KR-20 formula
kr_20 = (k_mkb / (k_mkb - 1)) * (1 - (sum_pq_mkb / sigma_t_squared_mkb))

print(f"Kuder-Richardson Formula 20 (KR-20): {kr_20:.1f}")
```

Using the KR-20 formula, the internal consistency is **0.7** which can be interpreted as *acceptable*.

### Cronbach's Alpha Reliability Test on Pre-Matched Belief Dataset

```python
# Using the mbb1 dataset
# 1. Number of items (k)
k_mbb = mbb1.shape[1]
print(f"Number of items (k) for mbb1: {k_mbb}")
```

```python
# 2. Calculate the variance of each item (sigma_i^2) and sum them up
sum_sigma_i_squared_mbb = mbb1.var().sum()
print(f"Sum of item variances (sum_sigma_i^2) for mbb1: {sum_sigma_i_squared_mbb:.1f}")
```

```python
# 3. Calculate variance of total scores (sigma_t^2)
# Calculate total score for each respondent (sum across rows)
total_scores_mbb = mbb1.sum(axis=1)

# Calculate the variance of these total scores
sigma_t_squared_mbb = np.var(total_scores_mbb)
print(f"Variance of total scores (sigma_t^2) for mbb1: {sigma_t_squared_mbb:.1f}")
```

```python
# 4. Apply the Cronbach's Alpha formula
cronbach_alpha_mbb = (k_mbb / (k_mbb - 1)) * (1 - (sum_sigma_i_squared_mbb / sigma_t_squared_mbb))
print(f"Cronbach's Alpha for mbb1: {cronbach_alpha_mbb:.1f}")
```

The internal consistency of data in the Pre-Matched belief is **0.5**. This can be interpreted as *poor/ unacceptable*.
